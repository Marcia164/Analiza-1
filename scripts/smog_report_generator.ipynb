```python
import datetime as dt
from nbconvert.exporters import HTMLExporter
from nbconvert import PDFExporter
import nbformat
from nbconvert.preprocessors import ExecutePreprocessor
```

### Set variables

```python
raport_ts = dt.datetime.now().strftime("%Y%m%d_%H-%M")
output_html = f"../reports/smog_html{raport_ts}.html"
output_pdf = f"../reports/smog_pdf{raport_ts}.pdf"
formatted_notebook = nbformat.read('smog_raw.ipynb', as_version=4)
```

### Execute notebook 

```python
executor = ExecutePreprocessor(timeout=600, kernel_name="python3")
executor.preprocess(formatted_notebook)
```

### Export notebook to HTML

```python
# Get HTML data and exclude input cells

html_exporter = HTMLExporter()
html_exporter.exclude_input = True
(notebook_body, notebook_resources) = html_exporter.from_notebook_node(formatted_notebook)

# Save HTML data into the file

with open(output_html,'w', encoding='utf-8-sig') as f:
    f.write(notebook_body)
```

### Export notebook to pdf

```python
# Get PDF data and exclude input cells

pdf_exporter = PDFExporter()
pdf_exporter.exclude_input = True
(pdf_body, pdf_resources) = pdf_exporter.from_notebook_node(formatted_notebook)

# Save PDF data into the file

with open(output_pdf, "wb") as f:
    f.write(pdf_body)
```
